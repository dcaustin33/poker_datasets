```python
import pandas as pd
import random
from pokerbench_translator import create_pokerkit_state_postflop
from hand_to_text import convert_hand_to_narratives

df = pd.read_csv("/Users/derek/Desktop/poker_datasets/datasets/postflop_500k_train_set_game_scenario_information.csv")


random_number = random.randint(0, len(df))

def get_all_preflop_actors(preflop_action: str):
    """
    Based on the preflop action this should return 
    all the actors in the preflop action
    """
    of_interest = ["SB", "BB", "UTG", "HJ", "CO", "BTN"]
    output = []
    for i in range(len(preflop_action.split("/")) - 1, -1, -1):
        if preflop_action.split("/")[i] in of_interest and preflop_action.split("/")[i] not in output:
            output.append(preflop_action.split("/")[i])
    return "/".join(output)

idx = random.randint(0, len(df) - 1)
# idx = 121496

print(df.iloc[idx]['preflop_action'])
print(df.iloc[idx]['postflop_action'])
print(df.iloc[idx]['correct_decision'])
print(df.iloc[idx]['hero_position'])
print(df.iloc[idx])

# print(df.iloc[idx].keys())
# print(df.iloc[idx])

hand_history = create_pokerkit_state_postflop(df.iloc[idx])
print(hand_history.actions)
narratives = convert_hand_to_narratives(hand_history)
for i in range(1, 7):

    if i in narratives:
        print(narratives[i])
        print("-"*100)
        break
```
