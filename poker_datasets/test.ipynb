# Check the postflop dataset

```python
import random

import pandas as pd
from hand_to_text import convert_hand_to_narratives
from pokerbench_translator import create_pokerkit_state_postflop
from verify_pokerkit_hand import create_state, verify_hand
import tqdm

df = pd.read_csv("/Users/derek/Desktop/poker_datasets/datasets/postflop_500k_train_set_game_scenario_information.csv")


random_number = random.randint(0, len(df))

issue_hands = []

for i in tqdm.tqdm(range(len(df))):
    idx = i
    hand_history = create_pokerkit_state_postflop(df.iloc[idx])
    if not verify_hand(hand_history.actions, hand_history.blinds_or_straddles[0], hand_history.starting_stacks):
        issue_hands.append(idx)


```

```python
# portion to check individual hands
idx = random.choice(issue_hands)
print(df.iloc[idx]['preflop_action'])
print(df.iloc[idx]['postflop_action'])
print(df.iloc[idx]['correct_decision'])
print(df.iloc[idx]['hero_position'])
print(df.iloc[idx])

hand_history = create_pokerkit_state_postflop(df.iloc[idx])
print(hand_history.actions)
print("\nHand verified", verify_hand(
    hand_history.actions, hand_history.blinds_or_straddles[0], hand_history.starting_stacks
), "\n")
# starting_stacks = hand
narratives = convert_hand_to_narratives(hand_history)
for i in range(1, 7):
    if i in narratives:
        print(narratives[i])
        print("-"*100)
        break
```

```python
import os

from hand_to_text import convert_hand_to_narratives
from pokerkit import HandHistory
from verify_pokerkit_hand import verify_hand

path_to_hands = "/Users/derek/Desktop/phh-dataset/data/pluribus"

# Walk through files and find .phh files

def walk_through_hands(path_to_hands):
    for root, dirs, files in os.walk(path_to_hands):
        for file in files:
            if file.endswith('.phh'):
                phh_path = os.path.join(root, file)
                
                # Read the hand history
                with open(phh_path, 'rb') as f:
                    hand_history = HandHistory.load(f)
                    
                # Verify the hand
                try:
                    verify_hand(hand_history.actions, hand_history.blinds_or_straddles[0], hand_history.starting_stacks)
                            
                except Exception as e:
                    print(f"Error verifying hand: {e}")
                    return
walk_through_hands(path_to_hands)
```
